# KaggleUI on Colab

Colab sessions are **interactive**: KaggleUI is not a way to keep a GPU session alive, and everything stops when the runtime disconnects.

1. **Runtime → Change runtime type → GPU** (T4 or better).
2. Add secrets via the **key icon** in the left sidebar: `HF_TOKEN`, optionally `NGROK_AUTHTOKEN`, and for
   **GitHub sync** `KAGGLEUI_WORKSPACE_REPO` (`owner/repo`) plus `GITHUB_TOKEN` (fine-grained, *Contents: read and write*).
3. Run the cell below and open the printed link.

## What differs from Kaggle

There is no `/kaggle/working` output, no `/kaggle/input` Datasets mount, and no attached-secrets panel: use Colab Secrets (key icon) instead. Model weights, virtualenvs and job logs live under `/tmp/kaggleui` and disappear on disconnect, so attach a storage connection or mount Google Drive manually before converting or training large models. The panel, APIs, engines, Model Studio and multi-model residency are identical to the Kaggle build.
